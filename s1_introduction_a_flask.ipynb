## Flask, un framework d'application Web

Flask est une **framework Python** pour le développement d'applications Web. Flask est conçu pour être facile d'utilisation et minimal. Flask est pensé pour être étendu avec des *plugins* qui ajoutent des fonctionnalités spécifiques (p.ex. interaction avec base de données).

À noter qu'il existe d'autres frameworks Web en Python, parmi lesquels:
- `Django`, qui offre beaucoup plus de fonctionnalités, mais qui est beaucoup plus complexe d'utilisation
- `FastAPI`, qui est fait pour la rapidité

On utilise Flask car il est particulièrement simple (et élégant !) à prendre en main.

### Installation

**Avant votre première utilisation**, lancer les commandes suivantes pour créer un environnement virtuel et installer les dépendances.

#### Rappel: les environnements virtuels 

Pour rappel, **un environnement virtuel** est comme une boîte à outils Python qui contient tout ce dont vous avez besoin pour un projet donné.  
- c'est un **dossier qui contient une version "isolée" de Python et les librairies** dont on a besoin pour un projet donné.
- tout ce qui est fait dans l'environnement **n'aura aucun impact sur votre Python "global"** (p.ex.: les paquets installés dans un environnement virtuel ne seront pas installés sur toute votre machine)
- notre environnement s'appellera **`.venv`**
- **quelques commandes utiles**:
    ```bash
    # créer un environnement
    python3 -m venv <nom-de-l'env>
    # sourcer/activer un environnement
    source <nom-de-l'env>/bin/activate
    # vérifier si on utilise le bon environnement
    which python  # doit être une version de Python contenue dans votre dossier .venv/
    # sortir d'un environnement
    deactivate
    ``` 

#### Créer un environnement pour le cours

# Une première application

## Lancer l'appli

**Lancez la première application** avec la commande suivante, puis ouvrez `127.0.0.1:5000` pour voir le résultat:

```bash
# rappelez vous de bien avoir sourcé votre environnement et installé les dépendances
# ctrl+c pour arrêter l'appli
python3 apps/s1/base/main.py
```

**Le code de l'application est vraiment tout simple**:

````python
from flask import Flask

app = Flask("Ma première appli")

@app.route("/")
def index():
    return "Hello world !"
````

Pour lancer l'application, on fait:

````python
# cette commande a déjà été lancée dans "main.py"
app.run(debug=True)
````

## Le code de l'application

Reprenons: voici code complet de l'application:

```python
from flask import Flask

app = Flask("Ma première appli")

@app.route("/")
def index():
    return "Hello world !"

if __name__ == "__main__":
    app.run()
```

Commentons le maintenant ligne-par-ligne:

1. **On importe la classe** `Flask`
    ```python
    from flask import Flask
    ```

2. **On crée un objet `app`** instance de `Flask`. 
    - cet objet contient notre appli Flask. 
    - donner le nom de l'application est obligatoire, puisqu'il permet à votre machine de différencier les applications Flask qui tournent
    ```python
    app = Flask("Ma première appli")
    ```

3. **On crée une route**. Une route est une fonction qui associe à un schéma d'URL une fonction qui défini comment les requêtes à cette URL seront traitées (ici, une requête à `/` affichera `Hello world !` sur votre page web)
    ```python
    @app.route("/")
    def index():
        return "Hello world !"
    ```

4. **On lance l'application**
    ```python
    if __name__ == "__main__":
        app.run()
    ```

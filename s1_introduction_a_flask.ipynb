## Flask, un framework d'application Web

Flask est une **framework Python** pour le développement d'applications Web. Flask est conçu pour être facile d'utilisation et minimal. Flask est pensé pour être étendu avec des *plugins* qui ajoutent des fonctionnalités spécifiques (p.ex. interaction avec base de données).

À noter qu'il existe d'autres frameworks Web en Python, parmi lesquels:
- `Django`, qui offre beaucoup plus de fonctionnalités, mais qui est beaucoup plus complexe d'utilisation
- `FastAPI`, qui est fait pour la rapidité

On utilise Flask car il est particulièrement simple (et élégant !) à prendre en main.

Pour résumer: en fonction de l'URL, votre appli Web va déterminer quel contenu afficher.

## Routes: une définition

Les routes permettent de **naviguer dans une application**. Une route associe:
- **un schéma d'URL**
- **une fonction** qui définit comment les requêtes à cette URL seront traitées.

Dans notre exemple, une requête à `/` (= la base de l'URL) affichera `Hello world !` dans votre navigateur web.

![schema de route basique](img/schema_route_1.png)

## Une route en Flask

Décortiquons notre exemple:

```py
@app.route("/")
def index():
    return "Hello world !"
```

Une route contient **deux parties**:
- `@app.route("/")`: définit le schéma d'URL
- `def index(): ...`: définit la fonction associée à ce schéma. **On appelle cette fonction une "vue" (ou `view`, ou `view function`)**. C'est le résultat de la vue qui sera affiché sur le navigateur.

=> **quand on accède à l'URL `/`, c'est la valeur retournée par la fonction (ou "vue") `index()` qui sera affichée sur le navigateur.**

### La syntaxe `@app.route`

**`@app.route` est un décorateur**:
- un élément de syntaxe Python **toujours précédé de `@` et suivi d'une fonction**.
- définition technique: un décorateur est une fonction qui modifie le comportement de la fonction qui le suit.
- **définition concrète (à retenir)**: en Flask, `@app.route` transforme la fonction suivante en une vue exécutée chaque fois que l'URL passée en paramètre est accédée.

**La syntaxe basique est: `@app.route("url_de_la_route")`**
- le `app` de `app.route` correspond au **nom de variable** de notre application Flask.
- **l'URL de la route une URL relative**: `/` correspond à la racine du site (`127.0.0.1:5000` dans notre navigateur)

## Routes à paramètres

Si votre application compte 1000 (ou 10000, ou 100000) pages, vous allez pas vous amuser à créer toutes ces routes à la main.

Vous allez plutôt **créer des routes à paramètres**, où 
- certaines parties sont constantes (ne changent jamais)
- certaines parties de l'URL sont définies dynamiquement
- ces parties sont passées comme des variables à votre fonction Python.

On lance `./apps/s1/routes_parametres/main.py`, qui définit la route suivante:

````python
@app.route("/page/<id_page>")
def chemin_page(id_page):
    return "Vous êtes sur la page " + id_page
````

### Syntaxe

En Flask, on définit la part variable d'une URL:
- dans `@app.route` encadrée de `<>`: `<ma_variable>`
- dans la fonction (ou "vue") associée, on passe `ma_variable` en argument

```python
@app.route("/page/<ma_variable>")
def chemin_page(ma_variable):
    ...
```

![schéma de route à paramètres](img/schema_route_2.png)

### Types

On peut typer les paramètres d'URL **pour qu'une URL soit valide seulement si le paramètre est d'un certain type**.


````python
@app.route("/page/<int:id_page>")
def chemin_page_type(id_page: int):
    return "Vous êtes sur la page " + id_page
````

**La syntaxe** est, dans `@app.route`: `mon_type:ma_variable`.

Voici tous les types définis par Flask:


type   | définition | exemple
-------|------------|--------
string | (défaut) n'importe quelle chaîne de caractères | `foo`
int	   | nombres entiers | `1`
float  | nombres décimaux | `5.5`
path   | chemin de fichiers | `/path/to/foo.txt`
uuid   | [UUID](https://fr.wikipedia.org/wiki/Universally_unique_identifier) (identifiant unique) | `123e4567-e89b-12d3-a456-426614174000` 

C'est très utile:
- **les paramètres seront passés à la vue avec le bon type** (si votre paramètre est `<int:id_page>`, alors dans votre fonction `chemin_page_type`, l'argument `id_page` sera déjà converti en `int`)
- **pour vous (développeur.euse)**, typer permet de documenter mieux vos données, donc de rendre votre code plus compréhensible (c'est TRÈS important)

## En résumé: un site calculatrice

Pour finir le cours, résumons avec un exemple de route plus complexe: **une route-calculatrice**, qui permet de **faire des calculs entre deux nombres**.

**Pour voir le résultat**, lancez `./apps/s1/routes_calculatrice/main.py`. Testez les URLs:
- [`http://localhost:5000/addition/2.5/2.5`](http://localhost:5000/addition/2.5/2.5)
- [`http://localhost:5000/multiplication/2.5/2.5`](http://localhost:5000/multiplication/2.5/2.5)
- [`http://localhost:5000/division/2.5/0.0`](http://localhost:5000/division/2.5/0.0)

### 1. La fonction `calculatrice` Python

En Python, on a besoin de 3 arguments:
- `operation`: le type d'opération à faire (`addition`, `soustraction`, `multiplication`, `division`)
- `x`: le premier nombre à calculer
- `y`: le deuxième nombre à calculer

Le code de notre calculatrice pourrait s'exprimer avec la fonction `calculatrice`:

````python
# seu;es ces opérations sont possibles
operations_autorisees = ["addition", "soustraction", "multiplication", "division"]

# cette fonction transforme le résultat en une chaîne de caractères
def format_reponse(operation: str, resultat: float) -> str:
    return "Le résultat de votre " + operation + " est : " + str(resultat) 

# notre fonction calculatrice
def calculatrice(operation: str, x: float, y: float) -> str:
    # on vérifie que `operation` est bien une opération autorisée
    if operation not in operations_autorisees:
        return "Manant ! Cette opération est interdite : " + operation + ". Les opérations autorisées sont : " + str(operations_autorisees)
    # on calcule le résultat 
    if operation == "addition": 
        resultat = x + y
    elif operation == "soustraction": 
        resultat = x - y
    elif operation == "multiplication": 
        resultat = x * y
    elif operation == "division":
        # la division par 0 est impossible => on gère cette erreur
        if y == 0:
            return "Manant ! La division par 0 est impossible !"
        resultat = x / y
    # on retourne une chaîne de caractère qui décrive ce résultat
    return format_reponse(operation, resultat)

# on teste les "bonnes résponses"
print(calculatrice("addition", 3.5, 1.5))
print(calculatrice("soustraction", 3.5, 1.5))
print(calculatrice("multiplication", 3.5, 1.5))
print(calculatrice("division", 4, 2))

# on teste les "erreurs"
print(calculatrice("division", 3.5, 0))
print(calculatrice("blah", 3.5, 4.8))
````

### 2. Une route Flask qui appelle `calculatrice`

Il faut maintenant connecter cette calculatrice à une route Flask. Pour ce faire, il suffit d'associer `calculatrice` à un `@app.route`:

````python
# on crée la route calculatrice.
# `route_calculatrice` est une vue Flask qui appelle la fonction `calculatrice`, définie ci-dessus.
# la fonction `calculatrice` se charge de calculer le résultat
# ici on voit que l'anatomie de notre URL suit exactement les arguments attendus par la fonction
# quelques limitations à noter: 
#   - x et y doivent être des floats, pas des entiers ! 
#       `/addition/1/2` ne marchera pas, `/addition/1.0/1.0` si
#   - x et y doivent être des nombres positifs
@app.route("/<string:operation>/<float:x>/<float:y>")
def route_calculatrice(operation: str, x: float, y: float):
    return calculatrice(operation, x, y)
````
